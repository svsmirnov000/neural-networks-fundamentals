# Inversion Matrix

Write code that generates a matrix that reverses the order of coordinates of a vector in $N$-dimensional space:

$$
\begin{bmatrix}
x_1 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_N \\
\vdots \\
x_1
\end{bmatrix}
$$

```python
def inversion_matrix(N):
    res = np.zeros([N, N])
    res[np.arange(N), np.arange(N-1, -1, -1)] = 1

    return res
```

```python
# Let us check, how does it work

x = np.arange(5)
print('x:', x)
print('Ax:', inversion_matrix(5) @ x)
```

```python
answer['task1'] = []
for one_input in inputs['task1']:
    answer['task1'].append(inversion_matrix(**one_input))
```

# Rolling Sum Matrix

$$
\begin{bmatrix}

x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_1 \\

\end{bmatrix}
$$

```python
def rolling_sum_matrix(N):
    res = np.zeros([N, N])
    for i in range(N):
        res[i, :i+1] = 1
    ## YOUR CODE HERE

    return res
```

# Neighbors Difference Matrix

Write code that generates a matrix that computes the differences between adjacent elements of a vector in $N$-dimensional space:

$$
\begin{bmatrix}
x_1 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_2 - x_1 \\
\vdots \\
x_N - x_{N-1}
\end{bmatrix}
$$

```python

def neighbors_difference_matrix(N):
    res = np.zeros((N - 1, N))
    np.fill_diagonal(res, -1)
    np.fill_diagonal(res[:, 1:], 1)
    ## YOUR CODE HERE
    return res
```

# Permutation Matrix

You are given a permutation $\sigma$, which specifies the order in which the elements should appear.

Write a program that generates a matrix that permutes
the coordinates of a vector:

$$
\begin{bmatrix}
x_1 \\
x_2 \\
\vdots \\
x_N
\end{bmatrix}
\rightarrow
A
\rightarrow
\begin{bmatrix}
x_{\sigma_1} \\
x_{\sigma_2} \\
\vdots \\
x_{\sigma_N}
\end{bmatrix}
$$

For example, if you are given a permutation $\sigma = (3, 4, 2, 1)$,
the resulting vector should have the following form:

$$[x_3, x_4, x_2, x_1]^T$$

```python
def permutation_matrix(order):
    N = len(order)
    res = np.zeros([N, N])
    for i, target in enumerate(order):
        col = target  # Убрали "-1", если индексы уже от 0 до N-1
        res[i, col] = 1
    return res
```

```python
# Let us check, how does it work

order = np.random.permutation(5)
x = np.arange(5)
print('x:', x)
print('order:', order)
print('Ax:', permutation_matrix(order) @ x)
```

```python
answer['task4'] = []
for one_input in inputs['task4']:
    answer['task4'].append(permutation_matrix(**one_input))
```

```python
def rotation_2D_matrix(phi):
#    res = np.zeros([2, 2])
    res = np.array([[np.cos(phi), -np.sin(phi)],
                    [np.sin(phi),  np.cos(phi)]])
    ## YOUR CODE HERE

    return res
```
