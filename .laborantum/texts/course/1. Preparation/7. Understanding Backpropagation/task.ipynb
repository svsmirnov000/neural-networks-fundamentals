# Task 1

Write a function that calculates the value of this graph for any given input vector `x` and set of coefficients $b_1, b_2, c_1, c_2$ packed as vector of weights `w`.
In this function also return all the $z$ values.

````python
import numpy as np

## YOU CAN DEFINE ANY FUNCTIONS YOU NEED ##
## YOUR CODE HERE ##

def sigmoid(t):
    return 1 / (1 + np.exp(-t))

def graph_value(x, w):
    y = x
    ## YOUR CODE HERE ##
    x1, x2 = x[0], x[1]
    b1, b2, c1, c2 = w[0], w[1], w[2], w[3]
    z1 = x1 + b1
    z2 = x2 + b2
    z3 = sigmoid(z1) 
    z4 = sigmoid(z2)
    z5 = np.tanh(z2)
    z6 = c2 * z5
    z7 = z1 * z4
    z8 = c1 * z7
    z9 = z3 * z6
    y = z8 + z9
    z = np.array([z1, z2, z3, z4, z5, z6, z7, z8, z9])
    
    ###end
    return y, z
````

````python
answer['graph_value'] = [graph_value(**input) for input in inputs]
````

# Task 2

In this graph, find all direct derivatives of all the values over $z$ or $x$ or $w$.

For example, if $y = z_8 + z_9$, you need to find only $\partial_{z_9} y$ and $\partial_{z_8} y$ for that operation.

You have to do that for all the intermediate signals.

Write the result in the form of a dictionary, for example, $\partial_{z_9} y$:

```
['dy_dz9'] = 1
```

````python
def graph_derivatives(x, w):
    res = {}
    ## YOUR CODE HERE ##
    res['dy_dz8'] = 1.0
    res['dy_dz9'] = 1.0
    
    
    x1, x2 = x[0], x[1]
    b1, b2, c1, c2 = w[0], w[1], w[2], w[3]
    z1 = x1 + b1
    z2 = x2 + b2
    z3 = sigmoid(z1)
    z4 = sigmoid(z2)
    z5 = np.tanh(z2)
    z6 = c2 * z5
    z7 = z1 * z4
    z8 = c1 * z7
    z9 = z3 * z6
    y = z8 + z9
    
    res['dz9_dz3'] = z6
    res['dz9_dz6'] = z3
    res['dz8_dc1'] = z7
    res['dz8_dz7'] = c1
    res['dz6_dc2'] = z5
    res['dz6_dz5'] = c2
    res['dz7_dz1'] = z4
    res['dz7_dz4'] = z1
    res['dz5_dz2'] = 1.0 - z5**2
    res['dz4_dz2'] = z4 * (1.0 - z4)
    res['dz3_dz1'] = z3 * (1.0 - z3)
    res['dz2_dx2'] = 1.0
    res['dz2_db2'] = 1.0
    res['dz1_dx1'] = 1.0
    res['dz1_db1'] = 1.0
    #end code

    return res
````

````python
answer['graph_derivatives'] = [graph_derivatives(**input) for input in inputs]
````

# Task 3

Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{c_1} y$
- calculate that derivative

````python
def dy_dc1(x, w):
    selected_derivs = {}
    dy_dc1 = 0

    ## YOUR CODE HERE ##
    y, z = graph_value(x, w)
    z7 = z[6]
    selected_derivs['dy_dz8'] = 1.0
    selected_derivs['dz8_dc1'] = z7
    dy_dc1 = selected_derivs['dy_dz8'] * selected_derivs['dz8_dc1']
    #end

    return selected_derivs, dy_dc1
````

````python
answer['dy_dc1'] = [dy_dc1(**input) for input in inputs]
````

# Task 4
Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{c_2} y$
- calculate that derivative

````python
def dy_dc2(x, w):
    selected_derivs = {}
    dy_dc2 = 0

    ## YOUR CODE HERE ##
    #y->z9->z6->c2
    y, z = graph_value(x, w)
    z3 = z[2]
    z5 = z[4]
    selected_derivs['dy_dz9'] = 1.0
    selected_derivs['dz9_dz6'] = z3
    selected_derivs['dz6_dc2'] = z5
    dy_dc2 = selected_derivs['dy_dz9'] * selected_derivs['dz9_dz6'] * selected_derivs['dz6_dc2']
    #end

    return selected_derivs, dy_dc2
````

````python
answer['dy_dc2'] = [dy_dc2(**input) for input in inputs]
````

# Task 5
Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{b_1} y$
- calculate that derivative

````python
def dy_db1(x, w):
    selected_derivs = {}
    dy_db1 = 0

    ## YOUR CODE HERE ##
    y, z = graph_value(x, w)
    z3 = z[2]
    z4 = z[3]
    z6 = z[5]
    b1, b2, c1, c2 = w[0], w[1], w[2], w[3]
    selected_derivs['dy_dz8'] = 1.0
    selected_derivs['dz8_dz7'] = c1
    selected_derivs['dz7_dz1'] = z4
    selected_derivs['dy_dz9'] = 1.0
    selected_derivs['dz9_dz3'] = z6
    selected_derivs['dz3_dz1'] = z3 * (1.0 - z3)  # Derivative of sigmoid(z1)
    selected_derivs['dz1_db1'] = 1.0
    up = selected_derivs['dy_dz8'] * selected_derivs['dz8_dz7'] * selected_derivs['dz7_dz1']
    low = selected_derivs['dy_dz9'] * selected_derivs['dz9_dz3'] * selected_derivs['dz3_dz1']
    dy_db1 = (up + low) * selected_derivs['dz1_db1']
    #end code

    return selected_derivs, dy_db1
````

````python
answer['dy_db1'] = [dy_db1(**input) for input in inputs]
````

# Task 6
Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{b_2} y$
- calculate that derivative

````python
def dy_db2(x, w):
    selected_derivs = {}
    dy_db2 = 0

    ## YOUR CODE HERE ##
    y, z = graph_value(x, w)
    z1 = z[0]
    z3 = z[2]
    z4 = z[3]
    z5 = z[4]
    b1, b2, c1, c2 = w[0], w[1], w[2], w[3]
    selected_derivs['dy_dz8'] = 1.0
    selected_derivs['dz8_dz7'] = c1
    selected_derivs['dz7_dz4'] = z1
    selected_derivs['dz4_dz2'] = z4 * (1.0 - z4)  # Derivative of sigmoid(z2)
    selected_derivs['dy_dz9'] = 1.0
    selected_derivs['dz9_dz6'] = z3
    selected_derivs['dz6_dz5'] = c2
    selected_derivs['dz5_dz2'] = 1.0 - z5**2      # Derivative of tanh(z2)
    selected_derivs['dz2_db2'] = 1.0
    up = (selected_derivs['dy_dz8'] * 
                  selected_derivs['dz8_dz7'] * 
                  selected_derivs['dz7_dz4'] * 
                  selected_derivs['dz4_dz2'])
    low = (selected_derivs['dy_dz9'] * 
                  selected_derivs['dz9_dz6'] * 
                  selected_derivs['dz6_dz5'] * 
                  selected_derivs['dz5_dz2'])
    
    dy_db2 = (up + low) * selected_derivs['dz2_db2']

    #end code

    return selected_derivs, dy_db2
````

````python
answer['dy_db2'] = [dy_db2(**input) for input in inputs]
````
